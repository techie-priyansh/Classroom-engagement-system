```python
import matplotlib.pyplot as plt
import numpy as np
import random

# 15 minutes total, with a data point every 10 seconds
# 15 * 60 = 900 seconds. Steps of 10 gives 91 points (including 0)
time_seconds = np.arange(0, 901, 10)
#engagement = np.zeros(len(time_seconds))
engagement = np.full(len(time_seconds), np.nan)
# Simulate a random walk for the non-zero points
current_value = 50.0  # Starting engagement level

for t in range(len(time_seconds)):
    # Plot a value for the first 10 seconds of a 30-second block (index 0, 3, 6...)
    if t % 30 < 10:
        # Simulate stock market-like random movement (-15 to +15)
        step = random.uniform(-15, 15)
        current_value += step
        
        # Clamp the value so it stays within a realistic 10-100 range for the peaks
        current_value = max(10, min(100, current_value))
        engagement[t] = current_value
    # else:
    #     # The next 20 seconds (2 data points) drop to 0
    #     engagement[t] = 0

# Set up the plot
plt.figure(figsize=(12, 5))
plt.plot(time_seconds, engagement, color='blue', linewidth=1.5, marker='.', markersize=5)

# Format the X-axis to show minutes (every 60 seconds)
x_ticks = np.arange(0, 901, 60)
x_labels = [f"{m}:00" for m in range(16)]
plt.xticks(x_ticks, x_labels)

# Configure axes and limits
plt.ylim(0, 100)
plt.xlim(0, 900)
plt.title("Student engagement- Student X")
plt.xlabel("Time (minutes)")
plt.ylabel("Engagement Level (0-100)")
plt.grid(True, linestyle='--', alpha=0.6)

# Display the plot
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import random

# 15 minutes total, with a data point EVERY SECOND
# 15 * 60 = 900 seconds. Steps of 1 gives 901 points (0 to 900)
time_seconds = np.arange(0, 901, 1)
engagement = np.zeros(len(time_seconds))

current_value = 50.0  # Starting engagement level

for t in time_seconds:
    # t % 30 gives the current second within the repeating 30-second block
    # If it's less than 10, we are in the "ON" period
    if t % 30 < 10:
        # Step size is smaller since it updates every second now
        step = random.uniform(-3, 3) 
        current_value += step
        
        # Clamp between 10 and 100 to keep it on the chart realistically
        current_value = max(10, min(100, current_value))
        engagement[t] = current_value
    else:
        # The remaining 20 seconds of the 30-second block are 0
        engagement[t] = 0

# Set up the plot
plt.figure(figsize=(14, 5))
# Using a thinner line and no markers so the 1-second data isn't cluttered
plt.plot(time_seconds, engagement, color='blue', linewidth=1.2)

# Format the X-axis to show minutes (ticks every 60 seconds)
x_ticks = np.arange(0, 901, 60)
x_labels = [f"{m}:00" for m in range(16)]
plt.xticks(x_ticks, x_labels)

# Configure axes
plt.ylim(0, 100)
plt.xlim(0, 120)
plt.title("Student engagement- Student X (1-Second Resolution)")
plt.xlabel("Time (minutes)")
plt.ylabel("Engagement Level (0-100)")
plt.grid(True, linestyle='--', alpha=0.6)

# Display the plot
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import random

# 15 minutes total, with a data point EVERY SECOND
time_seconds = np.arange(0, 901, 1)

# Initialize the array with NaN (Not a Number) instead of 0
# Matplotlib will not draw any lines connecting across NaN values
engagement = np.full(len(time_seconds), np.nan)

current_value = 50.0  # Starting engagement level

for t in range(len(time_seconds)):
    # If it's less than 10, we are in the "ON" period
    if t % 30 < 10:
        step = random.uniform(-3, 3) 
        current_value += step
        
        # Clamp between 10 and 100
        current_value = max(10, min(100, current_value))
        engagement[t] = current_value
    # No 'else' block needed; the value is already np.nan

# Set up the plot
plt.figure(figsize=(14, 5))
plt.plot(time_seconds, engagement, color='blue', linewidth=1.5)

# Format the X-axis to show minutes (ticks every 60 seconds)
x_ticks = np.arange(0, 901, 60)
x_labels = [f"{m}:00" for m in range(16)]
plt.xticks(x_ticks, x_labels)

# Configure axes
plt.ylim(0, 100)
plt.xlim(0, 120)
plt.title("Student engagement- Student X (Active Periods Only)")
plt.xlabel("Time (minutes)")
plt.ylabel("Engagement Level (0-100)")
plt.grid(True, linestyle='--', alpha=0.6)

# Display the plot
plt.tight_layout()
plt.show()
```
